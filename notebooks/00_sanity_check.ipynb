```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path 

spy=pd.read_parquet(r"C:\Users\Admin\Desktop\project_004\data\raw\spy_year_data.parquet")

print(spy.head())
print(spy.tail())

print(f"\nrows: {len(spy)}")
print(f"range: {spy.index.min().date()} -> {spy.index.max().date()}")
print(f"nulls:\n{spy.isna().sum()}")

gaps = spy.index.to_series().diff().dt.days
print(f"\nlargest gap: {gaps.max():.0f} days")

fig,ax = plt.subplots(2,1, figsize=(12,8),sharex=True)
ax[0].plot(spy.index,spy["Close"], linewidth=0.8)
ax[0].set_title("SPY close (unadjusted)")
ax[1].plot(spy.index, spy["volume"], linewidth=0.5, color="grey")
ax[1].set_title("Volume")
plt.tight_layout()
plt.show()
```
